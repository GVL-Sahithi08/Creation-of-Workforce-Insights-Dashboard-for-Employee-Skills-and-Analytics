**Prompt Used**: Using the existing Employees table in SQLite, perform basic workforce analysis by finding the total number of employees, average salary, highest salary, lowest salary, average experience, department-wise employee count, employees working overtime, and employees with attrition. Display the results using pandas in Google Colab.

```python
# Total Employees
print("Total Employees")
display(pd.read_sql("""
SELECT COUNT(*) AS TotalEmployees
FROM Employees
""", conn))

# Average Salary
print("Average Salary")
display(pd.read_sql("""
SELECT AVG(Salary) AS AverageSalary
FROM Employees
""", conn))

# Highest Salary
print("Highest Salary")
display(pd.read_sql("""
SELECT MAX(Salary) AS HighestSalary
FROM Employees
""", conn))

# Lowest Salary
print("Lowest Salary")
display(pd.read_sql("""
SELECT MIN(Salary) AS LowestSalary
FROM Employees
""", conn))

# Average Experience
print("Average Experience")
display(pd.read_sql("""
SELECT ROUND(AVG(Experience),2) AS AverageExperience
FROM Employees
""", conn))

# Department-wise Employee Count
print("Department-wise Employee Count")
display(pd.read_sql("""
SELECT Department,
       COUNT(*) AS EmployeeCount
FROM Employees
GROUP BY Department
""", conn))

# Employees Working Overtime
print("Employees Working Overtime")
display(pd.read_sql("""
SELECT EmployeeID,
       EmployeeName,
       Department
FROM Employees
WHERE Overtime='Yes'
""", conn))

# Employees with Attrition
print("Employees with Attrition")
display(pd.read_sql("""
SELECT EmployeeID,
       EmployeeName,
       Department
FROM Employees
WHERE Attrition='Yes'
""", conn))
```

**Prompt Used** : Using the existing Employees table in SQLite, perform advanced workforce analysis by retrieving the top 5 highest-paid employees, calculating the average salary for each department, counting employees based on their performance rating, and displaying employees with more than 5 years of experience. Present the results using SQL queries and display them using pandas in Google Colab.

```python
# Top 5 Highest-Paid Employees
print("Top 5 Highest-Paid Employees")
display(pd.read_sql("""
SELECT EmployeeID, EmployeeName, Department, Salary
FROM Employees
ORDER BY Salary DESC
LIMIT 5
""", conn))

# Employee Count by Performance Rating
print("Employee Count by Performance Rating")
display(pd.read_sql("""
SELECT PerformanceRating, COUNT(*) AS EmployeeCount
FROM Employees
GROUP BY PerformanceRating
""", conn))

# Average Salary by Department
print("Average Salary by Department")
display(pd.read_sql("""
SELECT Department,
ROUND(AVG(Salary),2) AS AverageSalary
FROM Employees
GROUP BY Department
ORDER BY AverageSalary DESC
""", conn))

# Employees with More Than 5 Years of Experience
print("Employees with More Than 5 Years of Experience")
display(pd.read_sql("""
SELECT EmployeeID,
EmployeeName,
Department,
Experience
FROM Employees
WHERE Experience > 5
ORDER BY Experience DESC
""", conn))
```
