**Prompt Used**: Using the previously created workforce dataset, perform basic workforce analytics using Python and pandas. Display the total number of employees, average salary, highest salary, lowest salary, average experience, department-wise employee count, employees working overtime, and employees with attrition. Present each result clearly with appropriate labels.

```python
# Total Employees
print("Total Employees:", len(df))

# Average Salary
print("\nAverage Salary:", round(df["Salary"].mean(), 2))

# Highest Salary
print("\nHighest Salary:", df["Salary"].max())

# Lowest Salary
print("\nLowest Salary:", df["Salary"].min())

# Average Experience
print("\nAverage Experience:", round(df["Experience"].mean(), 2), "Years")

# Department-wise Employee Count
print("\nDepartment-wise Employee Count:")
print(df["Department"].value_counts())

# Employees Working Overtime
print("\nEmployees Working Overtime:")
print(df[df["Overtime"] == "Yes"][["EmployeeID", "EmployeeName", "Department"]])

# Employees with Attrition
print("\nEmployees with Attrition:")
print(df[df["Attrition"] == "Yes"][["EmployeeID", "EmployeeName", "Department"]])
```

**Prompt Used** : Using the existing workforce dataset, perform advanced workforce analysis by displaying the top 5 highest-paid employees, counting employees based on performance rating, calculating the average salary for each department, and identifying employees with more than 5 years of experience. Display all results in a clear and organized format using pandas.

```python
# Top 5 Highest-Paid Employees
print("Top 5 Highest-Paid Employees:")
print(df.nlargest(5, "Salary")[["EmployeeID", "EmployeeName", "Department", "Salary"]])

# Count of Employees by Performance Rating
print("\nEmployee Count by Performance Rating:")
print(df["Performance"].value_counts())

# Average Salary by Department
print("\nAverage Salary by Department:")
print(df.groupby("Department")["Salary"].mean().round(2))

# Employees with More Than 5 Years of Experience
print("\nEmployees with More Than 5 Years of Experience:")
print(df[df["Experience"] > 5][["EmployeeID", "EmployeeName", "Department", "Experience"]])
```
